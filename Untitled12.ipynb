```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def analisar_vigor_otimizado(caminho_imagem):
    # 1. Carregar a imagem e converter para escala contínua (float)
    img = cv2.imread(caminho_imagem).astype(float) / 255.0
    B, G, R = cv2.split(img)

    # 2. Cálculo do VARI com offset no denominador para estabilização
    # Adiciona-se uma constante (0.1) para conter amplificações indevidas causadas pelo solo
    vari = (G - R) / (G + R - B + 0.1)

    # 3. Truncamento/Clipping de limites (Limita o índice na faixa real esperada de -0.5 a 0.5)
    # Evita que pixels extremos de brilho ou solo distorçam o contraste da folha
    vari_clipped = np.clip(vari, -0.5, 0.5)

    # 4. Reescalonamento para exibição visual (0 a 255)
    vari_norm = cv2.normalize(vari_clipped, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    # 5. Gerar o mapa de cores térmico
    heatmap = cv2.applyColorMap(vari_norm, cv2.COLORMAP_JET)

    # 6. Renderização gráfica no Matplotlib
    plt.figure(figsize=(14, 7))

    plt.subplot(1, 2, 1)
    plt.title("Imagem Original")
    plt.imshow(cv2.cvtColor((img * 255).astype(np.uint8), cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title("VARI Ajustado (Com Proteção de Solo)")
    plt.imshow(cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.show()

    print(f"Índice de Vigor Médio Ajustado: {np.mean(vari_clipped):.4f}")

# Execução
# analisar_vigor_otimizado('teste.jpeg')
```
